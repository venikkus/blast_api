```python
import sys
n = input()


def main():
    data = input().splitlines()
    n = int(data[0])
    moves = data[1:]

    from_counter = {}
    to_counter = {}

    for move in moves:
        from_addr, to_addr = ' '.join(move.split()[:2]), ' '.join(move.split()[2:])
        from_counter[from_addr] = from_counter.get(from_addr, 0) + 1
        to_counter[to_addr] = to_counter.get(to_addr, 0) + 1

    start, end = None, None

    addresses = set(from_counter.keys()) | set(to_counter.keys())

    for addr in addresses:
        from_count = from_counter.get(addr, 0)
        to_count = to_counter.get(addr, 0)

        if from_count - to_count == 1:
            if start is not None:
                print(-1)
                return
            start = addr
        elif to_count - from_count == 1:
            if end is not None:
                print(-1)
                return
            end = addr
        elif from_count != to_count:
            print(-1)
            return

    if start and end:
        print(start, end)
    else:
        print(-1)
```
